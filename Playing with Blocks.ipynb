```python
def invert_grid(grid_in):
    """
    Create a new grid with the same parameters as the passed
    in grid that has its colors inverted.

    """
    # Step 1: create new grid to return
    grid_out = BlockGrid(
        grid_in.width,
        grid_in.height,
        block_size=grid_in.block_size,
        lines_on=grid_in.lines_on
    )

    # Step 2: apply inversion to all blocks in grid_out
    for block in grid_in:
        red = block.red
        green = block.green
        blue = block.blue

        inverted = (255 - red, 255 - green, 255 - blue)

        grid_out[block.row, block.col].rgb = inverted

    # Step 3: return the inverted grid
    return grid_out
```

```python
invert_grid(grid)
```

```python
def invert_color(rgb):
    return tuple(255 - value for value in rgb)

def invert_grid2(grid_in):
    grid_out = grid_in.copy()

    for block_in, block_out in zip(grid_in, grid_out):
        block_out.rgb = invert_color(block_in.rgb)

    return grid_out
```

```python
def add_border(grid_in, border_color=colors.Black, border_size=1):
    # Step 1: create larger output grid
    grid_out = BlockGrid(
        grid_in.height + (2 * border_size),
        grid_in + (2 * border_size),
        fill=border_color,
        block_size=grid_in.block_size,
        lines_on=grid_in.lines_on
    )

    # Step 2: copy input grid to center of output grid
    grid_out[border_size:-border_size, border_size:-border_size] = grid_in

    return grid_out
```
